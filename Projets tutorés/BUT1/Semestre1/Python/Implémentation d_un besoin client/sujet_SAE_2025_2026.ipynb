Concrètement, on décide de stocker les réponses au questionnaire dans un tableau de chaînes de caractères et d'entiers contenant successivement un élève (sous forme d'une chaîne de caractères), et les quatres réponses aux questions sous forme d'entiers. Le nom de chaque élève se trouve à la case d'indice `[5*i]`, les réponses aux questions 1, 2, 3 et 4 se trouvent respectivement aux cases `[5 *i + 1]`, `[5 *i + 2]`, `[5 *i + 3]` et `[5 *i + 4]`. Un tel tableau est appelé *tableau de réponses*. On suppose qu'un tableau de réponses est toujours correctement construit (une chaîne de caractères pour chaque élève et 4 entiers entre 1 et 10 pour les réponses).

### Question 1

Définissez la fonction `nombre_eleves` prenant en paramètre un tableau de réponses et retournant le nombre d'élèves ayant répondu au questionnaire. Définissez également une fonction de tests unitaires.

````python
#Question n°1

""" Cette fonction prends en paramètre un tableau de réponses et retourne le nombre d'élèves ayant
répondu au questionnaire"""

def nombre_eleves (tab_reponses): #un tableau de réponses
    return (len(tab_reponses) //5) #car case d'indice [5*i]

def test_nombre_eleves():
    assert nombre_eleves(["Harry Potter", 10, 5, 8, 1]) ==1 #Vérification de si la fonction nous renvoi 1
    assert nombre_eleves(["Harry Potter", 10, 5, 8, 1, "Cedric Diggory", 6, 7, 9, 4]) != 3 #Vérification que la fonction ne retourne pas une valeur autre que de 2
    assert nombre_eleves ( ["Cedric Diggory", 6, 7, 9, 4, "Drago Malefoy", 1, 3, 2, 10]) ==2 #Vérification de si la fonction nous renvoi 2
    print ("Tout est bon")

#-----main-----------------
test_nombre_eleves()
print(nombre_eleves(reponses))
````

### Question 2

Définissez la fonction `eleves` prenant en paramètre un tableau de réponses et retournant un tableau contenant les élèves (chaînes de caractères) ayant répondu aux questionnaire. L'appel de cette fonction avec le tableau de réponses donné en exemple doit retourner le tableau :
```python
["Harry Potter", "Cedric Diggory", "Drago Malefoy"]
```

Définissez une fonction de tests unitaires pour la fonction `eleves`.

````python
#Question n°2

"""Cette fonction prends en paramètre un tableau de réponses et retourne un tableau contenant les
élèves (chaînes de caractères) ayant répondu aux questionnaire"""

def eleves (tab_reponses):
    nbr_eleve = nombre_eleves(tab_reponses) #nombre d'élèves ayant répondu au questionnaire
    i=0
    noms=[]
    while i < nbr_eleve:
        noms.append(tab_reponses [5*i]) #car nom est a l'indice 5*i
        i+=1
    return noms


def test_eleves():
    assert eleves(["Harry Potter", 10, 5, 8, 1, "Cedric Diggory", 6, 7, 9, 4, "Drago Malefoy", 1, 3, 2, 10])==['Harry Potter', 'Cedric Diggory', 'Drago Malefoy'] #Vérification que la fonction retourne ces noms
    assert eleves(["Cedric Diggory", 6, 7, 9, 4, "Drago Malefoy", 1, 3, 2, 10]) == ['Cedric Diggory','Drago Malefoy'] #Vérification que la fonction retourne ces 2 noms
    assert (eleves(["Harry Potter", 10, 5, 8, 1, "Cedric Diggory", 6, 7, 9, 4])) != ['Drago Malefoy','Cedric Diggory']#Vérification que la fonction ne retourne pas cette liste
    print("Tout est bon")
#-----main-----------------
test_eleves()
print(eleves(reponses))
````

Chaque ligne a le format `nom:R1/R2/R3/R4` où `nom` correspond au nom de l'élève et `R1`, `R2`, `R3`, `R4` correspondent aux scores (réponses aux questions) de l'élève. Par exemple, l'appel de la fonction `lecture_reponses` avec un nom de fichier contenant 
```
Harry Potter:10/5/8/1

```

Définissez une fonction de tests unitaires pour la fonction `lecture_reponses`.

````python
#Question n°3

"""Cette fonction prends en paramètre un nom de fichier et retourne un tableau de réponses en
fonction des données contenues dans le fichier"""

def lecture_reponses (nom_fichier):
    tab_reponses=[]
    f=open(nom_fichier)
    
    for ligne in f : #parcours chaque ligne du fichier
        
        ligne=ligne.strip()
        nom_score= ligne.split(":") #sépare nom et score
        nom= nom_score[0] #nom de l'élève
        scores= nom_score[1].split("/") #liste des scores
        
        tab_reponses.append(nom) #ajout du nom et de R1, R2, R3 et R4
        tab_reponses.append(int(scores[0])) # int car R1, R2, R3 et R4 sont des entiers 
        tab_reponses.append(int(scores[1]))
        tab_reponses.append(int(scores[2]))
        tab_reponses.append(int(scores[3]))
        
    f.close()
    
    return tab_reponses



def test_lecture_reponses ():
    
    fichier= "exemple.txt"
    f=open(fichier, "w") # Création d'un fichier temporaire
    f.write("Harry Potter:10/5/8/1\n") #chaque ligne contient le nom de l'élève suivi de ses 4 réponses au questionnaire
    f.write("Cedric Diggory:4/9/6/3\n")
    f.write("Drago Malefoy:1/2/3/10\n")
    f.close()
    resultat = ["Harry Potter", 10, 5, 8, 1,"Cedric Diggory", 4, 9, 6, 3,"Drago Malefoy", 1, 2, 3, 10]
    assert lecture_reponses(fichier)== resultat #Vérification de si la fonction nous renvoi le bon résultat
    print("Tout est bon")
    
#-----main-----------------
test_lecture_reponses ()
d=lecture_reponses("questionnaire_premiere_annee.txt")
print(d)
````

### Question 4

Définissez la fonction `maison` prenant en paramètre un tableau de réponses ainsi que l'indice d'un élève. La fonction retourne la maison correspondant à l'affectation de cet élève. Ainsi, l'appel de la fonction maison avec le tableau de réponses donné en exemple doit retourner : 
- `"Gryffondor"` avec l'indice `0` (correspondant à Harry Potter),
- `"Poufsouffle"` avec l'indice `5` (correspondant à Cedric Diggory),
- `"Serpentard"` avec l'indice `10` (correspondant à Drago Malefoy).

En effet, Harry Potter a mis le meilleur score à la question 1 et doit donc être affecté à Gryffondor. Cedric Diggory a mis le meilleur score à la question 3 et doit être affecté à la maison Poufsouffle. Finalement, Drago a mis le meilleur score à la question 4 et doit donc être affecté à la maison Serpentard.

Définissez également une fonction de tests unitaires pour la fonction `maison`.

````python
#Question n°4

"""Cette fonction prends en paramètre un tableau de réponses ainsi que l'indice d'un élève et
retourne la maison correspondant à l'affectation de cet élève"""

def maison(tab_reponses, indice):
    
    while indice%5 !=0: #Vérification de si l'indice donné correspond au debut d'un élève
        indice=int(input("saisissez un indice correct (5*i)"))
    i=1
    max= tab_reponses[indice+1]
    question= 1 #quel question possède la valeur la plus grande
    while i < 5:
        if tab_reponses[indice+i] > max :
            max =tab_reponses[indice+i] #recherche de la plus grande valeur
            question=i
        i+=1
        
    if question ==1: #Attribution de la maison en fonction du max
        return "Gryffondor"
    
    elif question==2:
        return "Serdaigle"
    
    elif question==3:
        return "Poufsouffle"
    
    else:
        return "Serpentard"
    
def test_maison() :
    
    reponses = ["Harry Potter", 10, 5, 8, 1, "Cedric Diggory", 6, 7, 9, 4, "Drago Malefoy", 1, 3, 2, 10]
    assert maison (reponses, 0) == "Gryffondor" #Vérification de si la fonction nous retourne "Gryffondor"
    assert maison (reponses, 5) != "Gryffondor" #Vérification que la fonction ne retourne pas lamauvaise maison
    assert maison (reponses, 10) == "Serpentard" #Vérification de si la fonction nous retourne "Serpentard"
    print("Tout est bon")
    
#-----main-----------------
test_maison()
print(maison(reponses, 0))
````

Définissez également une fonction de tests unitaires pour la fonction `repartition`.

````python
#Question n°5
"""Cette fonction prends en paramètre un tableau de réponses et retourne un dictionnaire dont les
clés sont les noms des élèves et les valeurs la maison à laquelle les élèves sont affectés"""

def repartition (tab_reponses):
    
    nbr_eleve = nombre_eleves(tab_reponses) #nombre d'élèves ayant répondu au questionnaire
    i=0
    dico ={}
    while i < nbr_eleve:
        indice_nom = 5*i #l'indice du nom de l'élève dans le tableau
        dico[eleves(tab_reponses)[i]]= maison(tab_reponses, indice_nom) #utilisation de la fonction eleves() pour récupérer le nom et maison() pour déterminer la maison
        i+=1
        
    return dico

def test_repartition():
    
    reponses = ["Harry Potter", 10, 5, 8, 1, "Cedric Diggory", 6, 7, 9, 4, "Drago Malefoy", 1, 3, 2, 10]
    assert repartition(reponses) == { "Harry Potter": "Gryffondor", "Cedric Diggory": "Poufsouffle","Drago Malefoy": "Serpentard"} #Vérification de si la fonction nous renvoi le bon dictionnaire
    assert not repartition(reponses) == { "Harry Potter": "Poufsouffle", "Cedric Diggory": "Gryffondor","Drago Malefoy": "Serpentard"} #Vérification de si la fonction nous renvoi pas le mauvais dictionnaire
    print("Tout est bon")
    
#-----main-----------------
test_repartition()
print(repartition(reponses))
````
